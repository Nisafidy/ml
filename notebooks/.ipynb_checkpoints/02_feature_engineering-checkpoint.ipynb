```python
import numpy as np
import pandas as pd

# 1. Chargement propre sans DtypeWarning
df_empl = pd.read_csv('../data/raw/EMPL_complet.csv', low_memory=False)
df_metier = pd.read_csv('../data/raw/metier.csv')
df_loge = pd.read_csv('../data/processed/df_eda_revenus.csv')

print(f'✅ EMPL_complet : {df_empl.shape[0]} individus')
print(f'✅ Logement (EDA) : {df_loge.shape[0]} ménages')

# 2. Harmonisation des identifiants (ex: idHH)
if 'idHH' in df_empl.columns and 'idHH_clean' in df_loge.columns:
    df_empl['idHH_clean'] = df_empl['idHH'].astype(str).str.zfill(12)

# ==============================================================================
# CONSTRUCTIONS DE LA CIBLE : score_individuel_epm
# ==============================================================================
# Remarque : Adapter les noms de colonnes exacts de ton EMPL_complet.csv si besoin.


def minmax_scale(series):
    """Normalise une série entre 0 et 1."""
    s_min, s_max = series.min(), series.max()
    if s_max == s_min:
        return series * 0
    return (series - s_min) / (s_max - s_min)


# --- Pilier 1 : Revenu (35%) ---
# Exemple avec une colonne de revenu (ex: q4a_68a ou salaire brut)
# On applique un log1p pour réduire l'asymétrie des hauts salaires
if 'revenu_mensuel' in df_empl.columns:
    rev_col = 'revenu_mensuel'
else:
    # Ajuster selon la colonne de salaire dans EMPL_complet.csv
    rev_col = [c for c in df_empl.columns if 'q4a_' in c or 'sal' in c][0]

df_empl['pilier_revenu'] = minmax_scale(
    np.log1p(df_empl[rev_col].fillna(0).clip(lower=0))
)

# --- Pilier 2 : Formalité de l'emploi (20%) ---
# Basé sur contrat, CNAPS, fiche de paie (variables EPM uniquement)
# On crée un score binaire ou catégoriel simple à titre d'illustration
df_empl['pilier_formalite'] = 0.5  # Valeur par défaut si non disponible

# --- Pilier 3 : Qualification (15%) ---
df_empl['pilier_qualification'] = 0.5

# --- Pilier 4 : Éducation (15%) ---
df_empl['pilier_education'] = 0.5

# --- Pilier 5 : Équipement / Services (15%) ---
df_empl['pilier_equipement'] = 0.5


# --- CALCUL DU SCORE FINAL (0 à 100) ---
df_empl['score_individuel_epm'] = (
    0.35 * df_empl['pilier_revenu']
    + 0.20 * df_empl['pilier_formalite']
    + 0.15 * df_empl['pilier_qualification']
    + 0.15 * df_empl['pilier_education']
    + 0.15 * df_empl['pilier_equipement']
) * 100

print('\n📊 Aperçu de la variable cible score_individuel_epm :')
print(df_empl['score_individuel_epm'].describe())
```

```python
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')

# 1. Chargement des données issues de l'EDA (01)
df_loge = pd.read_csv('../data/processed/df_eda_revenus.csv')

# 2. Chargement des données d'emploi brutes et du référentiel métiers
df_empl = pd.read_csv('../data/raw/EMPL_complet.csv', low_memory=False)
df_metier = pd.read_csv('../data/raw/metier.csv')

print(f'✅ Logement (EDA) : {len(df_loge)} ménages')
print(f'✅ Emploi (Individus) : {len(df_empl)} lignes')

# 3. Standardisation de la clé idHH sur 12 caractères
df_empl['idHH_clean'] = df_empl['idHH'].astype(str).str.zfill(12)
df_loge['idHH_clean'] = df_loge['idHH_clean'].astype(str).str.zfill(12)

# 4. Fusion pour récupérer l'environnement du ménage (milieu, région) sur chaque travailleur
df_merged = pd.merge(
    df_empl,
    df_loge[['idHH_clean', 'hhmilieu2', 'hhreg']],
    on='idHH_clean',
    how='inner',
)

print(f'✅ Fusion réussie : {len(df_merged)} travailleurs conservés')
```

```python
# Helper pour normaliser entre 0 et 1
def minmax_scale(series):
    s_min, s_max = series.min(), series.max()
    if s_max == s_min:
        return series * 0
    return (series - s_min) / (s_max - s_min)


# --- 1. PILIER REVENU (35%) ---
# Identification de la variable de salaire/revenu principal
rev_col = [c for c in df_merged.columns if 'q4a_68a' in c or 'sal' in c]
col_sal = rev_col[0] if rev_col else df_merged.columns[5]

df_merged['salaire_traite'] = (
    pd.to_numeric(df_merged[col_sal], errors='coerce').fillna(0).clip(lower=0)
)
df_merged['pilier_revenu'] = minmax_scale(
    np.log1p(df_merged['salaire_traite'])
)

# --- 2. PILIER FORMALITÉ DE L'EMPLOI (20%) ---
# Dictionnaires/règles selon présence de contrat, sécurité sociale (CNAPS, etc.)
formal_cols = [c for c in df_merged.columns if 'q4a_08' in c or 'q4a_12' in c]
if formal_cols:
    df_merged['pilier_formalite'] = minmax_scale(
        pd.to_numeric(df_merged[formal_cols[0]], errors='coerce').fillna(0)
    )
else:
    df_merged['pilier_formalite'] = 0.3  # Valeur par défaut si absente

# --- 3. PILIER QUALIFICATION (15%) ---
qualif_cols = [c for c in df_merged.columns if 'q4a_00' in c or 'q1_0x' in c]
if qualif_cols:
    df_merged['pilier_qualification'] = minmax_scale(
        pd.to_numeric(df_merged[qualif_cols[0]], errors='coerce').fillna(0)
    )
else:
    df_merged['pilier_qualification'] = 0.5

# --- 4. PILIER ÉDUCATION (15%) ---
edu_cols = [c for c in df_merged.columns if 'q11_' in c or 'edu' in c]
if edu_cols:
    df_merged['pilier_education'] = minmax_scale(
        pd.to_numeric(df_merged[edu_cols[0]], errors='coerce').fillna(0)
    )
else:
    df_merged['pilier_education'] = 0.4

# --- 5. PILIER ÉQUIPEMENT / ACCÈS SERVICES (15%) ---
df_merged['pilier_equipement'] = minmax_scale(
    df_merged['hhmilieu2'].astype(float)
)

# === CALCUL DU SCORE INDIVIDUEL EPM (0 à 100) ===
df_merged['score_individuel_epm'] = (
    0.35 * df_merged['pilier_revenu']
    + 0.20 * df_merged['pilier_formalite']
    + 0.15 * df_merged['pilier_qualification']
    + 0.15 * df_merged['pilier_education']
    + 0.15 * df_merged['pilier_equipement']
) * 100

print(
    '🎯 Variable cible score_individuel_epm créée avec succès !'
)
print(df_merged['score_individuel_epm'].describe())
```

```python
# 1. Sélection stricte des features de production
features_prod = [
    'idHH_clean',  # Clé d'agrégation ménage
    'salaire_traite',  # Salaire (déclaré ou estimé via metier.csv)
    'hhmilieu2',  # Milieu (Urbain / Rural)
    'hhreg',  # Région
]

# Ajout des identifiants métiers/secteurs disponibles dans EMPL_complet
metier_cols = [
    c
    for c in df_merged.columns
    if 'q4a_02' in c or 'metier' in c or 'branche' in c
]
if metier_cols:
    features_prod.append(metier_cols[0])

# Variable cible
col_target = ['score_individuel_epm']

# 2. Nettoyage de la matrice d'entraînement finale
df_dataset_ml = df_merged[features_prod + col_target].dropna(
    subset=['score_individuel_epm']
)

# Renommage propre des colonnes
df_dataset_ml.rename(
    columns={'salaire_traite': 'salaire_mensuel'}, inplace=True
)

# 3. Sauvegarde dans data/processed/
output_ml_path = '../data/processed/df_features_score.csv'
df_dataset_ml.to_csv(output_ml_path, index=False)

print(
    f'\n🚀 Dataset de modélisation prêt et exporté vers : {output_ml_path}'
)
print(
    f'📊 Dimensions finales : {df_dataset_ml.shape[0]} lignes × {df_dataset_ml.shape[1]} colonnes'
)
print('\nAperçu des 5 premières lignes du dataset d\'entraînement :')
df_dataset_ml.head()
```

```python
# ==============================================================================
# NETTOYAGE DES INDIVIDUS SANS MÉTIER ET AJUSTEMENT DU DATASET ML
# ==============================================================================

# 1. Filtrer les travailleurs ayant une valeur renseignée pour le métier (q4a_02)
df_ml_clean = df_dataset_ml.dropna(subset=['q4a_02']).copy()

# 2. Imputation propre du milieu / région si NaN
df_ml_clean['hhmilieu2'] = df_ml_clean['hhmilieu2'].fillna(
    df_ml_clean['hhmilieu2'].mode()[0]
)

# 3. Conversion du code métier en catégorie string/int propre
df_ml_clean['q4a_02'] = df_ml_clean['q4a_02'].astype(str)

# 4. Exportation du dataset filtré
output_ml_path = '../data/processed/df_features_score.csv'
df_ml_clean.to_csv(output_ml_path, index=False)

print(
    f'✅ Dataset nettoyé exporté : {df_ml_clean.shape[0]} travailleurs actifs'
)
print('📊 Distribution révisée de score_individuel_epm :')
print(df_ml_clean['score_individuel_epm'].describe())
```

```python
# Affichage des modalités uniques pour hhmilieu2 et hhreg

print("--- Valeurs uniques pour 'hhmilieu2' ---")
if 'hhmilieu2' in df_ml_clean.columns:
    print(df_ml_clean['hhmilieu2'].value_counts(dropna=False))
else:
    print("La colonne 'hhmilieu2' n'est pas présente sous ce nom.")

print("\n--- Valeurs uniques pour 'hhreg' ---")
if 'hhreg' in df_ml_clean.columns:
    print(df_ml_clean['hhreg'].value_counts(dropna=False))
else:
    print("La colonne 'hhreg' n'est pas présente sous ce nom.")
```

```python
import os
import joblib
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
from xgboost import XGBRegressor


# 2. Séparation Features (X) et Cible (y)
X = df_ml_clean.drop(columns=['idHH_clean', 'score_individuel_epm'])
y = df_ml_clean['score_individuel_epm']

# Encodage One-Hot des variables catégorielles (métier, milieu, région)
X_encoded = pd.get_dummies(X, drop_first=True)

# 3. Split Train / Test (80% / 20%)
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42
)

print(f"✅ Matrice X : {X_encoded.shape[1]} variables après encodage")
print(f"✅ Apprentissage : {X_train.shape[0]} lignes | Test : {X_test.shape[0]} lignes\n")

# 4. Initialisation des modèles
models = {
    "Random Forest": RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1),
    "XGBoost": XGBRegressor(n_estimators=100, learning_rate=0.08, max_depth=6, random_state=42, n_jobs=-1)
}

# 5. Entraînement et évaluation
results = []

for name, model in models.items():
    print(f"⏳ Entraînement de {name}...")
    model.fit(X_train, y_train)
    
    y_pred = model.predict(X_test)
    
    mae = mean_absolute_error(y_test, y_pred)
    rmse = np.sqrt(mean_squared_error(y_test, y_pred))
    r2 = r2_score(y_test, y_pred)
    
    results.append({
        "Modèle": name,
        "MAE": round(mae, 4),
        "RMSE": round(rmse, 4),
        "R2 Score": round(r2, 4)
    })

# Affichage du tableau comparatif
df_results = pd.DataFrame(results)
print("\n📊 COMPARAISON DES PERFORMANCES DU MODULE 4 :")
print(df_results.to_string(index=False))
```

```python
# Sélection du meilleur modèle (ex: XGBoost)
best_model = models["XGBoost"]

# Création du dossier models si inexistant
os.makedirs('../models', exist_ok=True)

# Export de l'artefact .pkl
model_path = '../models/score_individuel_model.pkl'
joblib.dump(best_model, model_path)

print(f"\n💾 Meilleur modèle sauvegardé avec succès dans : {model_path}")
```

```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# 1. Chargement du modèle sauvegardé et des données
model = joblib.load('../models/score_individuel_model.pkl')
df = pd.read_csv('../data/processed/df_features_score.csv')

X = pd.get_dummies(
    df.drop(columns=['idHH_clean', 'score_individuel_epm']), drop_first=True
)
y = df['score_individuel_epm']

# 2. Prédictions globales
y_pred = model.predict(X)

# 3. Visualisation : Valeurs Réelles vs Prédictions & Distribution des Erreurs
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Graphique 1 : Scatter plot
sns.scatterplot(
    x=y, y=y_pred, ax=axes[0], alpha=0.3, color='#2b5c8f', edgecolor=None
)
axes[0].plot([y.min(), y.max()], [y.min(), y.max()], 'r--', lw=2)
axes[0].set_title('Valeurs Réelles vs Prédictions ML')
axes[0].set_xlabel('score_individuel_epm (Réel EPM)')
axes[0].set_ylabel('score_individuel_epm (Prédit ML)')

# Graphique 2 : Distribution des résidus (erreurs)
residus = y - y_pred
sns.histplot(residus, kde=True, ax=axes[1], color='#e74c3c', bins=30)
axes[1].set_title('Distribution des Erreurs (Résidus)')
axes[1].set_xlabel('Erreur (Réel - Prédit)')

plt.tight_layout()
plt.show()

print('✅ Analyse d\'évaluation générée avec succès.')
```

```python
import joblib
import numpy as np
import pandas as pd


class SocioEconomicScoreModule:
    """Module 4 du DSAA : Calcul du score individuel et du family_score."""

    def __init__(self, model_path: str = "models/score_individuel_model.pkl"):
        self.model = joblib.load(model_path)

    def predict_individual_score(self, worker_data: dict) -> float:
        """Predit le score_individuel_epm d'un travailleur à partir de ses caractéristiques."""
        df_input = pd.DataFrame([worker_data])
        # Alignement automatique des colonnes catégorielles lors de l'inférence
        df_encoded = pd.get_dummies(df_input)

        # Réalignement avec les colonnes attendues par le modèle
        model_features = getattr(
            self.model, "feature_names_in_", df_encoded.columns
        )
        df_encoded = df_encoded.reindex(
            columns=model_features, fill_value=0
        )

        score = float(self.model.predict(df_encoded)[0])
        return max(0.0, min(100.0, score))

    def compute_family_score(self, workers: list) -> float:
        """Calcule le family_score agrégé pondéré par les salaires des membres du ménage.

        Formule : family_score = sum(score_i * salaire_i) / sum(salaire_i)
        """
        if not workers:
            return 0.0

        total_weighted_score = 0.0
        total_salary = 0.0

        for w in workers:
            score_i = self.predict_individual_score(w)
            salary_i = w.get("salaire_mensuel", 0.0)

            total_weighted_score += score_i * salary_i
            total_salary += salary_i

        if total_salary == 0:
            # En cas de revenu nul déclaré, moyenne arithmétique simple
            return float(
                np.mean([
                    self.predict_individual_score(w) for w in workers
                ])
            )

        return float(total_weighted_score / total_salary)
```
