```python
import json
import joblib
import pandas as pd

# 1. Chargement du dataset d'apprentissage
df = pd.read_csv('../data/processed/df_features_score.csv')
X = df.drop(columns=['idHH_clean', 'score_individuel_epm'])
X_encoded = pd.get_dummies(X, drop_first=True)

# 2. Extraction de la liste exacte des colonnes d'entraînement
feature_columns = list(X_encoded.columns)

# 3. Sauvegarde dans ml/model_metadata.json
with open('../ml/model_metadata.json', 'w', encoding='utf-8') as f:
    json.dump({'feature_names': feature_columns}, f, ensure_ascii=False, indent=2)

print("✅ Fichier ml/model_metadata.json créé avec succès !")
print(f"📊 Nombre de colonnes d'entraînement enregistrées : {len(feature_columns)}")
```
