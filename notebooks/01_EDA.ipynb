```python
# 1. Charger le fichier métier
df_metier = pd.read_csv('../data/raw/metier.csv')

# Créer un dictionnaire métier -> salaire moyen
metier_map = dict(zip(df_metier['metier'].str.lower(), df_metier['salaire_moyen_ariary']))

# 2. Fonction de correspondance sur la profession (q4a_18)
def estimer_salaire_metier(row):
    prof = str(row.get('q4a_18', '')).lower().strip()
    
    if not prof or prof == 'nan':
        return 196359 # Valeur nationale par défaut
    
    # Recherche exacte ou partielle dans la liste des métiers
    for metier, salaire in metier_map.items():
        if metier in prof or prof in metier:
            return salaire
            
    # Rapprochements sectoriels par mots-clés clés
    if any(m in prof for m in ['cultiv', 'agri', 'femi', 'elev', 'paysan', 'planteur']):
        return 225000  # Ouvrier / exploitant agricole
    elif any(m in prof for m in ['mancov', 'manoeuv', 'journalier']):
        return 200000  # Ouvrier non qualifié
    elif any(m in prof for m in ['vend', 'commer', 'marcha']):
        return 325000  # Vendeur / commerce
    elif any(m in prof for m in ['chauff', 'driver', 'transp']):
        return 425000  # Chauffeur
    elif any(m in prof for m in ['macon', 'btp', 'charpen']):
        return 325000  # Maçon / BTP
    elif any(m in prof for m in ['enseig', 'insti', 'prof']):
        return 350000  # Éducation
    elif any(m in prof for m in ['agent', 'fonc', 'admin']):
        return 450000  # Administration
        
    return 196359

# 3. Application sur les actifs
actifs = df[df['q4a_17'].notnull()].copy()
actifs['salaire_impute_v2'] = actifs.apply(estimer_salaire_metier, axis=1)

print("--- NOUVELLE DISTRIBUTION DES SALAIRES IMPUTÉS ---")
print(actifs['salaire_impute_v2'].describe())
```

```python
col_id_menage = 'idHH'

# 1. Calcul de la somme des salaires imputés par ménage
revenu_par_menage = actifs.groupby(col_id_menage)['salaire_impute_v2'].agg(
    revenu_total_menage='sum',
    nb_actifs_menage='count'
).reset_index()

# 2. Fusion des indicateurs ménage dans le DataFrame principal (df)
df = df.merge(revenu_par_menage, on=col_id_menage, how='left')

# Remplacement des valeurs manquantes pour les ménages sans actif
df['revenu_total_menage'] = df['revenu_total_menage'].fillna(0)
df['nb_actifs_menage'] = df['nb_actifs_menage'].fillna(0)

# 3. Statistiques au niveau ménage (en filtrant les doublons)
df_menages = df.drop_duplicates(subset=[col_id_menage])

print("--- STATISTIQUES DES REVENUS PAR MÉNAGE ---")
print(f"Nombre total de ménages distincts : {len(df_menages)}")
print(f"Ménages avec au moins 1 actif : {(df_menages['nb_actifs_menage'] > 0).sum()}")
print("\nDistribution du revenu total par ménage (en Ariary) :")
print(df_menages[df_menages['revenu_total_menage'] > 0]['revenu_total_menage'].describe())
```

```python
# 1. Filtrer les ménages avec revenu > 0
df_menages_actifs = df[df['revenu_total_menage'] > 0].drop_duplicates(subset=['idHH']).copy()

# 2. Définir des limites de tranches (bins) adaptées aux revenus observés
# Min, 200k, 350k, 500k, 1M, Max
seuils = [0, 200000, 350000, 500000, 1000000, np.inf]
labels_tranches = ['Q1 - Très faible (< 200k)', 
                   'Q2 - Faible (200k-350k)', 
                   'Q3 - Moyen (350k-500k)', 
                   'Q4 - Élevé (500k-1M)', 
                   'Q5 - Très élevé (> 1M)']

# 3. Application du découpage avec pd.cut
df_menages_actifs['quintile_revenu'] = pd.cut(
    df_menages_actifs['revenu_total_menage'], 
    bins=seuils, 
    labels=labels_tranches
)

# 4. Répercuter le quintile dans le DataFrame principal df
# On supprime d'abord la colonne si elle existe déjà suite à un essai précédent
if 'quintile_revenu' in df.columns:
    df = df.drop(columns=['quintile_revenu'])

df = df.merge(df_menages_actifs[['idHH', 'quintile_revenu']], on='idHH', how='left')

# Les ménages sans revenu actif passent en "Sans revenu imposable/déclaré"
df['quintile_revenu'] = df['quintile_revenu'].astype(str).replace('nan', 'Non renseigné / Sans actif')

print("--- RÉPARTITION DES MÉNAGES PAR TRANCHE DE REVENU ---")
print(df.drop_duplicates(subset=['idHH'])['quintile_revenu'].value_counts())
```

```python
# 1. Préparation des données (1 ligne par ménage)
df_menages_unique = df.drop_duplicates(subset=['idHH']).copy()

# Ordre logique des tranches
ordre_tranches = [
    'Q1 - Très faible (< 200k)',
    'Q2 - Faible (200k-350k)',
    'Q3 - Moyen (350k-500k)',
    'Q4 - Élevé (500k-1M)',
    'Q5 - Très élevé (> 1M)',
]

# Calcul des effectifs
counts = (
    df_menages_unique['quintile_revenu']
    .value_counts()
    .reindex(ordre_tranches)
    .fillna(0)
)
total_menages = counts.sum()

# 2. Configuration du style
plt.figure(figsize=(12, 6))
sns.set_theme(style='whitegrid')

# Graphique en barres (syntaxe v0.13+ corrigée)
ax = sns.barplot(
    x=counts.index,
    y=counts.values,
    hue=counts.index,  # Requis avec palette
    palette='viridis',  # Minuscule
    legend=False,  # Supprime la légende inutile
)

# 3. Ajout des annotations (Effectifs + Pourcentages)
for p in ax.patches:
  height = p.get_height()
  if not np.isnan(height) and height > 0:
    percentage = f'{(height / total_menages) * 100:.1f}%'
    annotation = f'{int(height):,}\n({percentage})'
    ax.annotate(
        annotation,
        (p.get_x() + p.get_width() / 2.0, height),
        ha='center',
        va='bottom',
        fontsize=10,
        fontweight='bold',
        xytext=(0, 5),
        textcoords='offset points',
    )

# 4. Habillage du graphique
plt.title(
    'Répartition des ménages par tranche de revenu estimé (en Ariary)',
    fontsize=14,
    fontweight='bold',
    pad=20,
)
plt.xlabel('Tranche de revenu du ménage', fontsize=11, labelpad=10)
plt.ylabel('Nombre de ménages', fontsize=11)
plt.xticks(rotation=15, ha='right', fontsize=10)
plt.ylim(0, max(counts.values) * 1.15)

plt.tight_layout()
plt.show()
```

```python
# Fonction pour calculer les proportions pondérées
def taux_pondere(df, var_groupe, var_cible, poids='hhweight'):
    grouped = df.groupby([var_groupe, var_cible])[poids].sum().unstack()
    taux = grouped.div(grouped.sum(axis=1), axis=0) * 100
    return taux.round(2)

print("=== TAUX D'ACCÈS AU RÉSEAU ÉLECTRIQUE PAR MILIEU (%) ===")
# hhmilieu2 : 1 = Urbain, 2 = Rural (ou selon codage INSTAT)
print(taux_pondere(df_merged, 'hhmilieu2', 'target_reseau'))

print("\n=== TAUX D'ACCÈS À L'ÉCLAIRAGE PROPRE (RÉSEAU + SOLAIRE) PAR MILIEU (%) ===")
print(taux_pondere(df_merged, 'hhmilieu2', 'target_eclairage_propre'))
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# 1. Chargement
df_empl = pd.read_csv('../data/raw/EMPL_complet.csv', low_memory=False)
df_loge = pd.read_csv('../data/processed/df_eda_revenus.csv')

print("="*80)
print("CONSTRUCTION DU SCORE EPM - VERSION ADAPTÉE")
print("="*80)

# 2. Fusion des données (si idHH compatible)
# Nettoyer les ID
df_empl['idHH_clean'] = df_empl['idHH'].astype(str).str.zfill(12)
df_loge['idHH_clean'] = df_loge['idHH_clean'].astype(str).str.zfill(12)

# Fusion
df_merged = df_empl.merge(df_loge[['idHH_clean', 'hhmilieu2', 'q11_02a', 'q11_19', 'q11_20', 'q11_21', 
                                   'target_reseau', 'target_eclairage_propre']], 
                          on='idHH_clean', how='left')

print(f"✅ {len(df_merged)} individus après fusion")

# ==============================================================================
# PILIER 1 : REVENU (35%)
# ==============================================================================

print("\n📊 PILIER 1 : REVENU")

# Utiliser q4a_46b ou q4a_46c (salaire)
rev_cols = ['q4a_46b', 'q4a_46c']
df_merged['revenu_individuel'] = 0

for col in rev_cols:
    if col in df_merged.columns:
        # Remplacer les valeurs manquantes par 0
        df_merged[col] = df_merged[col].fillna(0)
        # Convertir en numérique si nécessaire
        df_merged[col] = pd.to_numeric(df_merged[col], errors='coerce').fillna(0)
        df_merged['revenu_individuel'] += df_merged[col]

# Log transformation
df_merged['pilier_revenu'] = np.log1p(df_merged['revenu_individuel'])

# Normalisation
scaler_rev = MinMaxScaler()
df_merged['pilier_revenu'] = scaler_rev.fit_transform(df_merged[['pilier_revenu']])

print(f"  ✅ Revenu : {df_merged['pilier_revenu'].mean():.3f} (moyenne normalisée)")
print(f"  📊 Distribution :")
print(df_merged['pilier_revenu'].describe())

# ==============================================================================
# PILIER 2 : FORMALITÉ (20%) - PROXY PAR LE LOGEMENT
# ==============================================================================

print("\n📊 PILIER 2 : FORMALITÉ (proxy via logement)")

# Utiliser le milieu et les équipements comme proxy du niveau de vie formel
df_merged['pilier_formalite'] = 0.5  # Valeur par défaut

# Bonus pour les ménages connectés au réseau
if 'target_reseau' in df_merged.columns:
    df_merged['pilier_formalite'] += df_merged['target_reseau'] * 0.3

# Bonus pour les ménages en milieu urbain
if 'hhmilieu2' in df_merged.columns:
    df_merged['pilier_formalite'] += (df_merged['hhmilieu2'] == 1).astype(int) * 0.2

# Normalisation
df_merged['pilier_formalite'] = MinMaxScaler().fit_transform(df_merged[['pilier_formalite']])

print(f"  ✅ Formalité : {df_merged['pilier_formalite'].mean():.3f} (moyenne normalisée)")

# ==============================================================================
# PILIER 3 : QUALIFICATION (15%) - PROXY PAR ÉDUCATION & LOGEMENT
# ==============================================================================

print("\n📊 PILIER 3 : QUALIFICATION")

# Proxy: nombre de pièces + qualité du logement
df_merged['pilier_qualification'] = 0.5

# Nombre de pièces
if 'q11_02a' in df_merged.columns:
    df_merged['q11_02a'] = df_merged['q11_02a'].fillna(df_merged['q11_02a'].median())
    df_merged['pilier_qualification'] += df_merged['q11_02a'] / df_merged['q11_02a'].max() * 0.3

# Matériaux du toit (proxy qualité)
if 'q11_20' in df_merged.columns:
    # Matériaux modernes = meilleure qualification
    toit_qualite = df_merged['q11_20'].isin([1, 2, 3, 4]).astype(int)  # Béton, tôle, etc.
    df_merged['pilier_qualification'] += toit_qualite * 0.2

# Normalisation
df_merged['pilier_qualification'] = MinMaxScaler().fit_transform(df_merged[['pilier_qualification']])

print(f"  ✅ Qualification : {df_merged['pilier_qualification'].mean():.3f} (moyenne normalisée)")

# ==============================================================================
# PILIER 4 : ÉDUCATION (15%) - PROXY PAR LOGEMENT
# ==============================================================================

print("\n📊 PILIER 4 : ÉDUCATION (proxy via équipements)")

# Proxy: accès à l'électricité et équipements
df_merged['pilier_education'] = 0.5

# Accès à l'éclairage propre
if 'target_eclairage_propre' in df_merged.columns:
    df_merged['pilier_education'] += df_merged['target_eclairage_propre'] * 0.3

# Milieu urbain (meilleur accès à l'éducation)
if 'hhmilieu2' in df_merged.columns:
    df_merged['pilier_education'] += (df_merged['hhmilieu2'] == 1).astype(int) * 0.2

# Normalisation
df_merged['pilier_education'] = MinMaxScaler().fit_transform(df_merged[['pilier_education']])

print(f"  ✅ Éducation : {df_merged['pilier_education'].mean():.3f} (moyenne normalisée)")

# ==============================================================================
# PILIER 5 : ÉQUIPEMENT (15%) - DIRECTEMENT DU LOGEMENT
# ==============================================================================

print("\n📊 PILIER 5 : ÉQUIPEMENT")

# Utiliser les variables d'équipement disponibles
df_merged['pilier_equipement'] = 0.5

# Réseau électrique
if 'target_reseau' in df_merged.columns:
    df_merged['pilier_equipement'] += df_merged['target_reseau'] * 0.3

# Éclairage propre
if 'target_eclairage_propre' in df_merged.columns:
    df_merged['pilier_equipement'] += df_merged['target_eclairage_propre'] * 0.2

# Normalisation
df_merged['pilier_equipement'] = MinMaxScaler().fit_transform(df_merged[['pilier_equipement']])

print(f"  ✅ Équipement : {df_merged['pilier_equipement'].mean():.3f} (moyenne normalisée)")

# ==============================================================================
# CALCUL DU SCORE FINAL
# ==============================================================================

print("\n" + "="*80)
print("📊 SCORE FINAL EPM")

df_merged['score_individuel_epm'] = (
    0.35 * df_merged['pilier_revenu'] +
    0.20 * df_merged['pilier_formalite'] +
    0.15 * df_merged['pilier_qualification'] +
    0.15 * df_merged['pilier_education'] +
    0.15 * df_merged['pilier_equipement']
) * 100

print("\n📈 Statistiques du score final :")
print(df_merged['score_individuel_epm'].describe())

print("\n📊 Distribution par milieu :")
if 'hhmilieu2' in df_merged.columns:
    milieu_labels = {1: 'Urbain', 2: 'Rural'}
    print(df_merged.groupby('hhmilieu2')['score_individuel_epm'].agg(['mean', 'std', 'count']))

# ==============================================================================
# VISUALISATION
# ==============================================================================

import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

piliers = ['pilier_revenu', 'pilier_formalite', 'pilier_qualification', 
           'pilier_education', 'pilier_equipement', 'score_individuel_epm']

for i, pilier in enumerate(piliers):
    if pilier in df_merged.columns:
        df_merged[pilier].hist(bins=50, ax=axes[i], alpha=0.7, edgecolor='black')
        axes[i].set_title(pilier)
        axes[i].set_xlabel('Valeur')
        axes[i].set_ylabel('Fréquence')
        axes[i].axvline(df_merged[pilier].mean(), color='red', linestyle='--', label=f'Moyenne: {df_merged[pilier].mean():.2f}')
        axes[i].legend()

plt.tight_layout()
plt.show()

# Sauvegarde
df_merged.to_csv('../data/processed/epm_score_complet.csv', index=False)
print("\n✅ Score sauvegardé dans 'epm_score_complet.csv'")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

print("="*80)
print("IMPUTATION DES SALAIRES PAR MÉTIER")
print("="*80)

# 1. Chargement des données
df_empl = pd.read_csv('../data/raw/EMPL_complet.csv', low_memory=False)
df_metier = pd.read_csv('../data/raw/metier.csv')
df_loge = pd.read_csv('../data/processed/df_eda_revenus.csv')

print(f"✅ EMPL_complet : {df_empl.shape[0]} individus")
print(f"✅ metier.csv : {df_metier.shape[0]} métiers")
print(f"✅ Logement : {df_loge.shape[0]} ménages")

# 2. Identifier les colonnes de métier dans EMPL_complet
print("\n" + "="*80)
print("IDENTIFICATION DES COLONNES DE MÉTIER DANS EMPL_complet")
print("="*80)

# Colonnes qui pourraient contenir des codes métier
metier_cols = [col for col in df_empl.columns if any(x in col.lower() for x in ['citp', 'nomac', 'q4a_19', 'q4a_23', 'activite'])]
print(f"Colonnes trouvées : {metier_cols}")

for col in metier_cols:
    print(f"\n📌 {col}:")
    print(f"  - Type: {df_empl[col].dtype}")
    print(f"  - Non-null: {df_empl[col].notna().sum()}")
    print(f"  - Unique: {df_empl[col].nunique()}")
    if df_empl[col].dtype == 'object' or df_empl[col].nunique() < 20:
        print(f"  - Exemples: {df_empl[col].dropna().head(10).tolist()}")
    else:
        print(f"  - Min: {df_empl[col].min()}, Max: {df_empl[col].max()}")

# 3. Nettoyage et préparation des colonnes de métier
print("\n" + "="*80)
print("PRÉPARATION DE LA CORRESPONDANCE MÉTIER")
print("="*80)

# Créer une colonne 'code_metier' standardisée
# Utiliser q4a_19 (CITP) comme code principal
if 'q4a_19' in df_empl.columns:
    df_empl['code_metier'] = df_empl['q4a_19'].astype(str).str.strip()
    print("✅ Utilisation de q4a_19 (CITP) comme code métier")

# Sinon utiliser q4a_23 (NOMAC)
elif 'q4a_23' in df_empl.columns:
    df_empl['code_metier'] = df_empl['q4a_23'].astype(str).str.strip()
    print("✅ Utilisation de q4a_23 (NOMAC) comme code métier")
else:
    print("⚠️ Aucune colonne de code métier trouvée")
    # Créer un code basé sur q4a_17 (nombre d'emplois)
    df_empl['code_metier'] = df_empl['q4a_17'].fillna(0).astype(int).astype(str)

# 4. Créer une table de correspondance secteur/métier
print("\n" + "="*80)
print("CRÉATION DE LA TABLE DE CORRESPONDANCE")
print("="*80)

# Extraire les codes des métiers du fichier metier.csv
def extraire_code_metier(nom_metier):
    """Extrait le code du métier depuis le nom"""
    if pd.isna(nom_metier):
        return None
    # Chercher des patterns comme 'CITP-1234' ou 'NOMAC-5678'
    import re
    match = re.search(r'(?:CITP|NOMAC)[\s-]*(\d+)', str(nom_metier))
    if match:
        return match.group(1)
    return None

df_metier['code_metier'] = df_metier['metier'].apply(extraire_code_metier)

# Afficher les codes trouvés
codes_trouves = df_metier[df_metier['code_metier'].notna()]
print(f"Codes extraits : {len(codes_trouves)} sur {len(df_metier)} métiers")

# 5. Fusionner les salaires par secteur (si pas de code direct)
print("\n" + "="*80)
print("FUSION DES SALAIRES PAR SECTEUR")
print("="*80)

# Créer une table des salaires moyens par secteur
salaire_par_secteur = df_metier.groupby('secteur').agg({
    'salaire_min_ariary': 'mean',
    'salaire_max_ariary': 'mean',
    'salaire_moyen_ariary': 'mean'
}).round(0).astype(int)

print("📊 Salaires moyens par secteur :")
print(salaire_par_secteur)

# 6. Imputation des salaires manquants
print("\n" + "="*80)
print("IMPUTATION DES SALAIRES")
print("="*80)

# Créer la colonne salaire_mensuel
df_empl['salaire_mensuel'] = 0

# Méthode 1 : Utiliser q4a_46c (salaire individuel)
if 'q4a_46c' in df_empl.columns:
    df_empl['salaire_mensuel'] = pd.to_numeric(df_empl['q4a_46c'], errors='coerce').fillna(0)
    print(f"✅ Salaire initial de q4a_46c : {df_empl['salaire_mensuel'].sum():.2f} total")

# Méthode 2 : Imputer par secteur (si disponible)
if 'q4a_23' in df_empl.columns:  # NOMAC
    # Extraire le code secteur (premiers chiffres)
    df_empl['secteur_code'] = df_empl['q4a_23'].astype(str).str[:2]
    
    # Créer une table de correspondance secteur -> salaire
    # Utiliser les données de metier.csv pour créer une correspondance
    correspondance_secteur = {}
    for idx, row in df_metier.iterrows():
        secteur = row['secteur']
        salaire_moyen = row['salaire_moyen_ariary']
        # Ajouter plusieurs variations du nom du secteur
        correspondance_secteur[secteur.lower()] = salaire_moyen
    
    # Fonction pour trouver le salaire moyen d'un secteur
    def get_salaire_secteur(row):
        # Si déjà un salaire > 0, le garder
        if row['salaire_mensuel'] > 0:
            return row['salaire_mensuel']
        
        # Sinon, essayer de trouver par secteur
        if 'secteur' in row and pd.notna(row['secteur']):
            secteur_lower = str(row['secteur']).lower()
            for key, value in correspondance_secteur.items():
                if key in secteur_lower or secteur_lower in key:
                    return value
        
        # Si toujours rien, utiliser la moyenne générale
        return df_metier['salaire_moyen_ariary'].median()
    
    # Appliquer l'imputation
    df_empl['salaire_mensuel'] = df_empl.apply(
        lambda row: get_salaire_secteur(row) if row['salaire_mensuel'] == 0 else row['salaire_mensuel'],
        axis=1
    )
    print(f"✅ Après imputation par secteur : {df_empl['salaire_mensuel'].sum():.2f} total")

# Méthode 3 : Si tout est encore à 0, utiliser la médiane par milieu
if df_empl['salaire_mensuel'].sum() == 0:
    print("⚠️ Tous les salaires sont à 0, utilisation de la médiane par milieu")
    
    # Fusionner avec df_loge pour avoir le milieu
    df_empl['idHH_clean'] = df_empl['idHH'].astype(str).str.zfill(12)
    df_loge['idHH_clean'] = df_loge['idHH_clean'].astype(str).str.zfill(12)
    
    df_empl = df_empl.merge(df_loge[['idHH_clean', 'hhmilieu2']], on='idHH_clean', how='left')
    
    # Salaire médian par milieu (urbain/rural)
    salaire_median_urbain = df_metier['salaire_moyen_ariary'].quantile(0.75)
    salaire_median_rural = df_metier['salaire_moyen_ariary'].quantile(0.25)
    
    df_empl['salaire_mensuel'] = np.where(
        df_empl['hhmilieu2'] == 1,  # Urbain
        salaire_median_urbain,
        salaire_median_rural
    )
    print(f"✅ Salaires imputés : Urbain={salaire_median_urbain:.0f}, Rural={salaire_median_rural:.0f}")

# 7. Statistiques finales des salaires
print("\n" + "="*80)
print("STATISTIQUES FINALES DES SALAIRES")
print("="*80)

print(f"📊 Distribution des salaires :")
print(df_empl['salaire_mensuel'].describe())

print(f"\n📊 Individus avec salaire > 0 : {(df_empl['salaire_mensuel'] > 0).sum()}")
print(f"📊 Pourcentage : {(df_empl['salaire_mensuel'] > 0).mean() * 100:.1f}%")

# 8. Création du pilier revenu
print("\n" + "="*80)
print("CRÉATION DU PILIER REVENU")
print("="*80)

# Normalisation log des salaires
df_empl['pilier_revenu'] = np.log1p(df_empl['salaire_mensuel'].clip(lower=0))
scaler_rev = MinMaxScaler()
df_empl['pilier_revenu'] = scaler_rev.fit_transform(df_empl[['pilier_revenu']])

print(f"✅ Pilier revenu créé :")
print(f"  - Moyenne: {df_empl['pilier_revenu'].mean():.3f}")
print(f"  - Min: {df_empl['pilier_revenu'].min():.3f}")
print(f"  - Max: {df_empl['pilier_revenu'].max():.3f}")

# 9. Préparation du fichier final pour le ML
print("\n" + "="*80)
print("PRÉPARATION DU FICHIER FINAL POUR LE ML")
print("="*80)

# Sélection des colonnes nécessaires
cols_final = ['idHH_clean', 'salaire_mensuel', 'hhmilieu2', 'hhreg', 'q4a_02', 'score_individuel_epm']

# S'assurer que toutes les colonnes existent
for col in cols_final:
    if col not in df_empl.columns:
        if col == 'hhreg' and 'hhreg' in df_loge.columns:
            df_empl = df_empl.merge(df_loge[['idHH_clean', 'hhreg']], on='idHH_clean', how='left')
        elif col == 'score_individuel_epm':
            # Recalculer le score si nécessaire
            df_empl['score_individuel_epm'] = df_empl['pilier_revenu'] * 100
        else:
            print(f"⚠️ Colonne {col} manquante, création avec valeur par défaut")
            df_empl[col] = 0

# Créer le fichier final
df_final = df_empl[cols_final].copy()

# Nettoyer les colonnes
df_final['hhmilieu2'] = df_final['hhmilieu2'].fillna(2)  # Défaut rural
df_final['hhreg'] = df_final['hhreg'].fillna(1)  # Défaut région 1
df_final['q4a_02'] = df_final['q4a_02'].fillna(0)

print(f"✅ Fichier final préparé : {df_final.shape}")
print(f"\n📋 Aperçu des données :")
print(df_final.head())

print(f"\n📊 Statistiques des variables finales :")
print(df_final.describe())

# 10. Sauvegarde
df_final.to_csv('../data/processed/final_ml_ready.csv', index=False)
print("\n✅ Fichier 'final_ml_ready.csv' sauvegardé !")

# 11. Visualisation
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Distribution des salaires
df_final['salaire_mensuel'].hist(bins=50, ax=axes[0,0], alpha=0.7, edgecolor='black')
axes[0,0].set_title('Distribution des salaires mensuels')
axes[0,0].set_xlabel('Salaire (Ariary)')
axes[0,0].set_ylabel('Fréquence')

# Distribution des scores
df_final['score_individuel_epm'].hist(bins=50, ax=axes[0,1], alpha=0.7, edgecolor='black', color='green')
axes[0,1].set_title('Distribution du score EPM')
axes[0,1].set_xlabel('Score')
axes[0,1].set_ylabel('Fréquence')

# Salaire par milieu
if 'hhmilieu2' in df_final.columns:
    df_final.boxplot(column='salaire_mensuel', by='hhmilieu2', ax=axes[1,0])
    axes[1,0].set_title('Salaire par milieu (1=Urbain, 2=Rural)')
    axes[1,0].set_xlabel('Milieu')
    axes[1,0].set_ylabel('Salaire')

# Score par milieu
df_final.boxplot(column='score_individuel_epm', by='hhmilieu2', ax=axes[1,1])
axes[1,1].set_title('Score EPM par milieu')
axes[1,1].set_xlabel('Milieu')
axes[1,1].set_ylabel('Score')

plt.tight_layout()
plt.show()

print("\n" + "="*80)
print("✅ PROCESSUS TERMINÉ !")
print("="*80)
```

```python
import pandas as pd
import numpy as np
from IPython.display import display, HTML

print("="*80)
print("COLONNES q4a_* AVEC DES VALEURS TEXTES NON BINAIRES")
print("="*80)

# Chargement
df_empl = pd.read_csv('../data/raw/EMPL_complet.csv', low_memory=False)

# Sélectionner toutes les colonnes q4a
q4a_cols = [col for col in df_empl.columns if col.startswith('q4a_')]

# Liste pour stocker les colonnes intéressantes
cols_interessantes = []

for col in q4a_cols:
    # Compter les valeurs non-nulles
    non_null = df_empl[col].notna().sum()
    if non_null == 0:
        continue
    
    # Récupérer les valeurs uniques (non-nulles)
    unique_vals = df_empl[col].dropna().unique()
    
    # Compter les valeurs uniques
    n_unique = len(unique_vals)
    
    # Vérifier si c'est binaire (Oui/Non, 1/0, True/False)
    is_binary = False
    if n_unique <= 2:
        # Vérifier si les valeurs sont binaires
        vals_lower = [str(v).lower().strip() for v in unique_vals if pd.notna(v)]
        binary_patterns = ['oui', 'non', 'yes', 'no', 'true', 'false', '1', '0', 'vrai', 'faux']
        if all(any(pattern in v for pattern in binary_patterns) for v in vals_lower):
            is_binary = True
    
    # Vérifier si c'est numérique
    is_numeric = False
    if df_empl[col].dtype in ['int64', 'float64']:
        is_numeric = True
    else:
        # Vérifier si toutes les valeurs peuvent être converties en nombre
        try:
            pd.to_numeric(df_empl[col].dropna())
            is_numeric = True
        except:
            pass
    
    # Garder seulement si c'est du texte non binaire
    if not is_binary and not is_numeric and n_unique > 2:
        # Calculer le pourcentage de remplissage
        pct = (non_null / len(df_empl)) * 100
        
        # Prendre un échantillon des valeurs
        sample_vals = unique_vals[:5].tolist()
        
        cols_interessantes.append({
            'Colonne': col,
            'Type': str(df_empl[col].dtype),
            '% Présent': round(pct, 1),
            'Valeurs uniques': n_unique,
            'Exemples': sample_vals,
            'Exemples (texte)': ' | '.join([str(v)[:30] for v in sample_vals])
        })

# Créer le DataFrame
df_result = pd.DataFrame(cols_interessantes)
df_result = df_result.sort_values('% Présent', ascending=False)

# ============================================================================
# 1. AFFICHAGE DU TABLEAU
# ============================================================================

print(f"\n✅ {len(df_result)} colonnes avec des valeurs textes non binaires\n")

if not df_result.empty:
    # Afficher le tableau complet
    display(df_result[['Colonne', '% Présent', 'Valeurs uniques', 'Exemples (texte)']])
    
    # ============================================================================
    # 2. AFFICHAGE SIMPLIFIÉ AVEC BARRES
    # ============================================================================
    
    print("\n" + "="*80)
    print("📋 AFFICHAGE SIMPLIFIÉ")
    print("="*80)
    
    for idx, row in df_result.iterrows():
        # Barre de progression
        bar_length = int(row['% Présent'] / 2)
        bar = '█' * bar_length + '░' * (50 - bar_length)
        
        # Extraire les exemples
        exemples = row['Exemples (texte)']
        if len(exemples) > 60:
            exemples = exemples[:60] + '...'
        
        print(f"\n📌 {row['Colonne']:<15}")
        print(f"   {bar} {row['% Présent']:>5.1f}%  |  {row['Valeurs uniques']:>3} valeurs")
        print(f"   Exemples: {exemples}")
    
    # ============================================================================
    # 3. SAUVEGARDE
    # ============================================================================
    
    df_result.to_csv('colonnes_texte_non_binaires.csv', sep=';', index=False, encoding='utf-8-sig')
    print("\n✅ Sauvegardé dans 'colonnes_texte_non_binaires.csv'")
    
else:
    print("❌ Aucune colonne avec des valeurs textes non binaires trouvée")

# ============================================================================
# 4. STATISTIQUES
# ============================================================================

print("\n" + "="*80)
print("📊 STATISTIQUES")
print("="*80)

# Compter par type
print(f"\nColonnes totales q4a : {len(q4a_cols)}")
print(f"Colonnes avec texte non binaire : {len(df_result)}")

if not df_result.empty:
    print(f"\nTop 10 colonnes par % de remplissage :")
    print(df_result.head(10)[['Colonne', '% Présent', 'Valeurs uniques']].to_string(index=False))
    
```
